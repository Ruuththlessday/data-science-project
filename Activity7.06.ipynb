```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import squarify
```

```python
data = pd.read_csv("data/age_salary_hours.csv")
```

```python
order = [
    "No diploma",
    "Attended college, no degree",
    "Associate's degree",
    "Bachelor's degree or higher",
    "High school diploma",
]

education_counts = data['Education'].value_counts().reindex(order)
education_percent = education_counts / education_counts.sum() * 100

labels = [
    f'{edu}\n({pct:.1f}%)'
    for edu, pct in zip(education_counts.index, education_percent)
]

plt.figure(dpi=200)

squarify.plot(
    sizes=education_percent,
    label=labels,
    color=sns.color_palette('colorblind', len(education_counts)),
    text_kwargs={'fontsize': 7}
)

plt.axis('off')
plt.title('Degrees')
plt.show()
```

```python
filtered_data = data[data['Age'] < 65]

fig, axes = plt.subplots(2, 1, figsize=(12, 10))

sns.violinplot(
    data=filtered_data,
    x='Education',
    y='Annual Salary',
    palette='colorblind',
    ax=axes[0]
)

sns.violinplot(
    data=filtered_data,
    x='Education',
    y='Weekly hours',
    palette='colorblind',
    ax=axes[1]
)

axes[0].set_title(
    'Impact of Education on Annual Salary and Weekly Working Hours'
)

axes[0].set_xlabel('Education')
axes[0].set_ylabel('Annual Salary')

axes[1].set_xlabel('Education')
axes[1].set_ylabel('Weekly Hours')

plt.tight_layout()
plt.show()
```
